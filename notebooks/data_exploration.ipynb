### Data Exploration

```python
import os
from pathlib import Path
import h5py
import matplotlib.pyplot as plt
import re
import numpy as np
```

Dataset files

```python
cwd = Path(os.getcwd())
files_dir = cwd.parent
files_dir = Path(files_dir, Path('output/modes'))

files_list = os.listdir(files_dir)
files_list = [file for file in files_list if file[-3:]=='.h5']
files_list = [Path(files_dir, Path(file)) for file in files_list]
files_list.sort()
```

Datasets classes exploration

```python
fig, ax = plt.subplots(2, 3, figsize=(8, 4))

for i in range(len(files_list)):
    hf = h5py.File(files_list[i], 'r')
    data = hf['images'][0]

    mode = str(files_list[i])
    mode = re.search(r"mode_l(\d+)_p(\d+)", mode)
    l = int(mode.group(1))
    p = int(mode.group(2))

    x_ind = i // 3
    y_ind = i % 3

    ax[x_ind, y_ind].imshow(data)
    ax[x_ind, y_ind].set_title(f'LG{l}{p}')

plt.tight_layout()
plt.show()
```
